### Target Leakage Correction

A sanity check identified a very strong correlation between `service_fee` and the target variable `price` (approximately 0.998).

Further inspection showed that the service fee is closely calculated from the Airbnb listing price. Therefore, including `service_fee` as a predictor would give the machine learning models indirect access to the target value and create target leakage.

The engineered variable `fee_to_price_ratio` was also excluded because it was calculated using the target price.

To ensure that model performance reflects genuine predictive relationships, both `service_fee` and `fee_to_price_ratio` were removed from the predictor variables before the models were retrained.

The dataset was divided into three subsets according to the project requirements:

## 4. Regression Model Training

Five regression algorithms were trained and compared to predict Airbnb listing prices:

1. Linear Regression
2. Decision Tree Regressor
3. Random Forest Regressor
4. Gradient Boosting Regressor
5. K-Nearest Neighbors Regressor

The models were trained using the 70% training dataset. Performance was first evaluated on the 15% validation dataset using Mean Absolute Error (MAE), Mean Squared Error (MSE), and R² Score.

The validation results are used to compare the models and identify the strongest candidates for ensemble modeling.

```python
# ======================================================
# Model Evaluation Function
# ======================================================

def evaluate_model(model_name, y_true, y_pred):
    
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    
    print("=" * 55)
    print(model_name)
    print("=" * 55)
    print(f"MAE: {mae:.4f}")
    print(f"MSE: {mse:.4f}")
    print(f"R² Score: {r2:.4f}")
    
    return {
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "R2": r2
    }

print("Evaluation function created successfully!")
```

### Regression Evaluation Metrics

Three evaluation metrics are used to compare model performance:

**Mean Absolute Error (MAE):** Measures the average absolute difference between the actual and predicted Airbnb prices. A lower MAE indicates better prediction accuracy.

**Mean Squared Error (MSE):** Measures the average squared difference between actual and predicted prices. Because larger errors are squared, MSE gives greater penalties to large prediction errors. A lower MSE is preferred.

**R² Score:** Measures the proportion of variation in Airbnb prices explained by the model. An R² score closer to 1 indicates stronger predictive performance.

Linear Regression serves as a baseline model for the project. It estimates the relationship between the predictor variables and Airbnb listing price using a linear equation.

### 4.2 Decision Tree Regressor

The Decision Tree Regressor predicts prices by repeatedly dividing the dataset into smaller groups based on feature values. Unlike Linear Regression, it can capture nonlinear relationships between Airbnb characteristics and price.

Decision trees do not require standardized features, so we will be using the original training and validation data.

Random Forest is an ensemble learning algorithm that combines predictions from multiple decision trees. Each tree is trained using different samples and subsets of features, and their predictions are averaged.

This approach will help reduce the overfitting associated with a single decision tree and can capture complex nonlinear relationships involve in the data.

```python
# ======================================================
# Model 3: Random Forest Regressor
# ======================================================

random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(
    X_train,
    y_train
)

random_forest_val_pred = random_forest_model.predict(
    X_val
)

random_forest_results = evaluate_model(
    "Random Forest Regressor",
    y_val,
    random_forest_val_pred
)

validation_results.append(random_forest_results)
```

```python
from sklearn.pipeline import Pipeline

print("Pipeline imported successfully!")
```

```python
# ======================================================
# Prepare Top 3 Models for Voting Regressor
# ======================================================

# Linear Regression pipeline
# Scaling occurs inside this model only.
voting_linear = Pipeline([
    ("scaler", StandardScaler()),
    ("linear", LinearRegression())
])

# Gradient Boosting
voting_gradient = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)

# Random Forest
voting_forest = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

print("Top three models prepared for ensemble.")
```

### 7.1 Voting Regressor

The Voting Regressor combines the predictions from Linear Regression, Gradient Boosting, and Random Forest. Each model contributes to the final prediction, and the individual predictions are averaged to produce the ensemble prediction.

Linear Regression is placed inside a preprocessing pipeline so that scaling is performed appropriately without unnecessarily scaling the tree-based models.

```python
# ======================================================
# Voting Regressor
# ======================================================

voting_model = VotingRegressor(
    estimators=[
        ("linear", voting_linear),
        ("gradient_boosting", voting_gradient),
        ("random_forest", voting_forest)
    ]
)

print("Training Voting Regressor...")

voting_model.fit(X_train, y_train)

print("Voting Regressor training completed!")
```

### 8.1 Bayesian-Weighted Ensemble

A Bayesian-style weighted ensemble was created to compare with the equal-weight Voting Regressor. Instead of giving the three selected models equal influence, model weights are derived from their validation performance.

Models with lower validation error receive greater weight in the final prediction. This allows stronger models to contribute more heavily while still incorporating information from the other selected models.

```python
# ======================================================
# Bayesian-Style Performance Weights
# ======================================================

top3_names = [
    "Linear Regression",
    "Gradient Boosting Regressor",
    "Random Forest Regressor"
]

top3_results = validation_comparison[
    validation_comparison["Model"].isin(top3_names)
].copy()

# Inverse MSE:
# smaller MSE produces a larger weight
top3_results["Inverse_MSE"] = 1 / top3_results["MSE"]

top3_results["Weight"] = (
    top3_results["Inverse_MSE"] /
    top3_results["Inverse_MSE"].sum()
)

print("Bayesian-style ensemble weights:")

display(
    top3_results[
        ["Model", "MSE", "Weight"]
    ]
)
```

```python
print(
    "Total weight:",
    top3_results["Weight"].sum()
)
```

```python
linear_weight = top3_results.loc[
    top3_results["Model"] == "Linear Regression",
    "Weight"
].iloc[0]

gradient_weight = top3_results.loc[
    top3_results["Model"] == "Gradient Boosting Regressor",
    "Weight"
].iloc[0]

forest_weight = top3_results.loc[
    top3_results["Model"] == "Random Forest Regressor",
    "Weight"
].iloc[0]

print("Linear Regression Weight:", round(linear_weight, 4))
print("Gradient Boosting Weight:", round(gradient_weight, 4))
print("Random Forest Weight:", round(forest_weight, 4))
```

```python
# ======================================================
# Bayesian-Weighted Validation Predictions
# ======================================================

bayesian_val_pred = (
    linear_weight * linear_val_pred
    + gradient_weight * gradient_boosting_val_pred
    + forest_weight * random_forest_val_pred
)

bayesian_val_results = evaluate_model(
    "Bayesian-Weighted Ensemble",
    y_val,
    bayesian_val_pred
)
```

```python
# ======================================================
# Bayesian-Weighted Test Predictions
# ======================================================

bayesian_test_pred = (
    linear_weight * linear_test_pred
    + gradient_weight * gradient_boosting_test_pred
    + forest_weight * random_forest_test_pred
)

bayesian_test_results = evaluate_model(
    "Bayesian-Weighted Ensemble - Test",
    y_test,
    bayesian_test_pred
)
```

Seven regression approaches were compared in this project: Linear Regression, Decision Tree Regressor, Random Forest Regressor, Gradient Boosting Regressor, K-Nearest Neighbors Regressor, Voting Regressor, and a Bayesian-weighted ensemble.

Linear Regression produced the strongest overall test performance. It achieved a test MAE of **2.5891**, MSE of **514.8252**, and R² of **0.9953**. This means that the model explained approximately 99.53% of the variation in the target variable within the test dataset.

### Gradient Boosting Regressor

Gradient Boosting was the second strongest individual model. It achieved a test MAE of **3.5011**, MSE of **518.2701**, and R² of **0.9953**. Its R² was very close to Linear Regression, but its MAE was higher. Therefore, Linear Regression produced smaller absolute prediction errors overall.

### Random Forest Regressor

Random Forest achieved a test MAE of **3.0790**, MSE of **599.6227**, and R² of **0.9945**. The model performed very well but did not outperform Linear Regression or Gradient Boosting. The results show that combining many decision trees reduced prediction error compared with using a single Decision Tree.

### Decision Tree Regressor

The Decision Tree achieved a test MAE of **3.4125**, MSE of **1049.5486**, and R² of **0.9904**. Although its R² remained high, its MSE was considerably larger than those of Linear Regression, Gradient Boosting, and Random Forest. This indicates that the Decision Tree produced some larger prediction errors.

### K-Nearest Neighbors Regressor

K-Nearest Neighbors produced the weakest results among the individual models. It achieved a test MAE of **82.6141**, MSE of **11429.5726**, and R² of **0.8954**. Although an R² of approximately 0.895 still indicates substantial explanatory ability, its prediction errors were much larger than those of the other models.

The weaker KNN performance suggests that similarity based on distance between observations was less effective for this dataset than the relationships learned by the other regression algorithms.

### Voting Regressor

The Voting Regressor combined Linear Regression, Gradient Boosting, and Random Forest. It achieved a test MAE of **2.8430**, MSE of **524.4012**, and R² of **0.9952**.

The ensemble performed extremely well but did not improve upon Linear Regression. Because voting averages predictions from several models, weaker predictions from Random Forest and Gradient Boosting slightly reduced the advantage provided by Linear Regression.

### Bayesian-Weighted Ensemble

The Bayesian-weighted ensemble assigned greater influence to models with stronger validation performance. It achieved a test MAE of **2.8393**, MSE of **523.3520**, and R² of **0.9952**.

It slightly outperformed the equal-weight Voting Regressor in MAE and MSE, showing that performance-based weighting provided a small improvement. However, it still did not outperform Linear Regression.

### Best Model

Based on the final test results, **Linear Regression was selected as the best-performing model**.

Its final test performance was:

- **MAE:** 2.5891
- **MSE:** 514.8252
- **R²:** 0.995287

Linear Regression achieved the lowest test MAE and MSE while also producing the highest test R². The results indicate that the relationships between the selected predictor variables and Airbnb price were captured very effectively by the linear model.

### Ensemble Comparison

The ensemble models did not outperform the strongest individual model. The Voting Regressor achieved an R² of approximately 0.9952, while the Bayesian-weighted ensemble also achieved approximately 0.9952. Linear Regression achieved the highest test R² of approximately 0.9953.

This result demonstrates that ensemble learning does not automatically guarantee better performance. An ensemble is most useful when its component models make different types of errors that can offset one another. In this project, Linear Regression was already highly accurate, so combining its predictions with slightly weaker models did not produce an improvement.

The Bayesian-weighted ensemble performed slightly better than the equal-weight Voting Regressor in MAE and MSE because stronger models received greater influence. However, the improvement was not enough to exceed Linear Regression.

# FINAL CORRECTED MODELING RESULTS

The following analysis replaces the earlier model results after correcting target leakage. The variables `service_fee` and `fee_to_price_ratio` were excluded because they contained information directly or indirectly related to the target variable `price`.

```python
# ============================================================
# FINAL CORRECTED MACHINE LEARNING ANALYSIS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression, BayesianRidge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 1. CORRECT FEATURES AND TARGET
# ------------------------------------------------------------

y_final = df["price"].copy()

X_final = df.drop(
    columns=[
        "price",
        "service_fee",
        "fee_to_price_ratio"
    ],
    errors="ignore"
).copy()

# Convert Boolean column to numeric
X_final["instant_bookable"] = X_final["instant_bookable"].astype(int)

print("Target leakage variables removed.")
print("Number of features:", X_final.shape[1])
print("Non-numeric columns:",
      X_final.select_dtypes(exclude=np.number).columns.tolist())
print("Missing values:", X_final.isnull().sum().sum())


# ------------------------------------------------------------
# 2. 70% TRAIN / 15% VALIDATION / 15% TEST
# ------------------------------------------------------------

X_train_f, X_temp_f, y_train_f, y_temp_f = train_test_split(
    X_final,
    y_final,
    test_size=0.30,
    random_state=42
)

X_val_f, X_test_f, y_val_f, y_test_f = train_test_split(
    X_temp_f,
    y_temp_f,
    test_size=0.50,
    random_state=42
)

total_f = len(X_final)

print("\nDATA SPLIT")
print("Training:",
      len(X_train_f),
      f"({len(X_train_f)/total_f:.1%})")
print("Validation:",
      len(X_val_f),
      f"({len(X_val_f)/total_f:.1%})")
print("Test:",
      len(X_test_f),
      f"({len(X_test_f)/total_f:.1%})")


# ------------------------------------------------------------
# 3. FEATURE SCALING
# ------------------------------------------------------------

scaler_f = StandardScaler()

X_train_scaled_f = scaler_f.fit_transform(X_train_f)
X_val_scaled_f = scaler_f.transform(X_val_f)
X_test_scaled_f = scaler_f.transform(X_test_f)


# ------------------------------------------------------------
# 4. TRAIN FIVE REQUIRED MODELS
# ------------------------------------------------------------

# Linear Regression
linear_f = LinearRegression()
linear_f.fit(X_train_scaled_f, y_train_f)

linear_val_f = linear_f.predict(X_val_scaled_f)
linear_test_f = linear_f.predict(X_test_scaled_f)

print("\n1/5 Linear Regression completed.")


# Decision Tree
tree_f = DecisionTreeRegressor(
    random_state=42
)

tree_f.fit(X_train_f, y_train_f)

tree_val_f = tree_f.predict(X_val_f)
tree_test_f = tree_f.predict(X_test_f)

print("2/5 Decision Tree completed.")


# Random Forest
forest_f = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

forest_f.fit(X_train_f, y_train_f)

forest_val_f = forest_f.predict(X_val_f)
forest_test_f = forest_f.predict(X_test_f)

print("3/5 Random Forest completed.")


# Gradient Boosting
gradient_f = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)

gradient_f.fit(X_train_f, y_train_f)

gradient_val_f = gradient_f.predict(X_val_f)
gradient_test_f = gradient_f.predict(X_test_f)

print("4/5 Gradient Boosting completed.")


# KNN
knn_f = KNeighborsRegressor(
    n_neighbors=5,
    n_jobs=-1
)

knn_f.fit(X_train_scaled_f, y_train_f)

knn_val_f = knn_f.predict(X_val_scaled_f)
knn_test_f = knn_f.predict(X_test_scaled_f)

print("5/5 KNN completed.")


# ------------------------------------------------------------
# 5. INDIVIDUAL MODEL VALIDATION RESULTS
# ------------------------------------------------------------

prediction_lookup_val = {
    "Linear Regression": linear_val_f,
    "Decision Tree Regressor": tree_val_f,
    "Random Forest Regressor": forest_val_f,
    "Gradient Boosting Regressor": gradient_val_f,
    "K-Nearest Neighbors Regressor": knn_val_f
}

prediction_lookup_test = {
    "Linear Regression": linear_test_f,
    "Decision Tree Regressor": tree_test_f,
    "Random Forest Regressor": forest_test_f,
    "Gradient Boosting Regressor": gradient_test_f,
    "K-Nearest Neighbors Regressor": knn_test_f
}

rows = []

for model_name in prediction_lookup_val:

    val_pred = prediction_lookup_val[model_name]
    test_pred = prediction_lookup_test[model_name]

    rows.append({
        "Model": model_name,

        "MAE_Validation":
            mean_absolute_error(y_val_f, val_pred),

        "MSE_Validation":
            mean_squared_error(y_val_f, val_pred),

        "R2_Validation":
            r2_score(y_val_f, val_pred),

        "MAE_Test":
            mean_absolute_error(y_test_f, test_pred),

        "MSE_Test":
            mean_squared_error(y_test_f, test_pred),

        "R2_Test":
            r2_score(y_test_f, test_pred)
    })


individual_results_f = pd.DataFrame(rows)


# ------------------------------------------------------------
# 6. FIND TOP 3 USING VALIDATION PERFORMANCE
# ------------------------------------------------------------

top3_f = individual_results_f.sort_values(
    by="R2_Validation",
    ascending=False
).head(3)

top3_names_f = top3_f["Model"].tolist()

print("\nTOP 3 MODELS FOR ENSEMBLE")

for name in top3_names_f:
    print("-", name)


# ------------------------------------------------------------
# 7. AVERAGE ENSEMBLE OF TOP 3
# ------------------------------------------------------------

ensemble_val_f = np.mean(
    [
        prediction_lookup_val[top3_names_f[0]],
        prediction_lookup_val[top3_names_f[1]],
        prediction_lookup_val[top3_names_f[2]]
    ],
    axis=0
)

ensemble_test_f = np.mean(
    [
        prediction_lookup_test[top3_names_f[0]],
        prediction_lookup_test[top3_names_f[1]],
        prediction_lookup_test[top3_names_f[2]]
    ],
    axis=0
)

ensemble_row_f = {
    "Model": "Average Ensemble (Top 3)",

    "MAE_Validation":
        mean_absolute_error(y_val_f, ensemble_val_f),

    "MSE_Validation":
        mean_squared_error(y_val_f, ensemble_val_f),

    "R2_Validation":
        r2_score(y_val_f, ensemble_val_f),

    "MAE_Test":
        mean_absolute_error(y_test_f, ensemble_test_f),

    "MSE_Test":
        mean_squared_error(y_test_f, ensemble_test_f),

    "R2_Test":
        r2_score(y_test_f, ensemble_test_f)
}


# ------------------------------------------------------------
# 8. BAYESIAN RIDGE MODEL
# ------------------------------------------------------------

bayesian_f = BayesianRidge()

bayesian_f.fit(
    X_train_scaled_f,
    y_train_f
)

bayesian_val_f = bayesian_f.predict(
    X_val_scaled_f
)

bayesian_test_f = bayesian_f.predict(
    X_test_scaled_f
)

bayesian_row_f = {
    "Model": "Bayesian Ridge",

    "MAE_Validation":
        mean_absolute_error(y_val_f, bayesian_val_f),

    "MSE_Validation":
        mean_squared_error(y_val_f, bayesian_val_f),

    "R2_Validation":
        r2_score(y_val_f, bayesian_val_f),

    "MAE_Test":
        mean_absolute_error(y_test_f, bayesian_test_f),

    "MSE_Test":
        mean_squared_error(y_test_f, bayesian_test_f),

    "R2_Test":
        r2_score(y_test_f, bayesian_test_f)
}


# ------------------------------------------------------------
# 9. FINAL VALIDATION + TEST COMPARISON
# ------------------------------------------------------------

final_corrected_comparison = pd.concat(
    [
        individual_results_f,
        pd.DataFrame([ensemble_row_f]),
        pd.DataFrame([bayesian_row_f])
    ],
    ignore_index=True
)

final_corrected_comparison = (
    final_corrected_comparison
    .sort_values(
        by="R2_Test",
        ascending=False
    )
    .reset_index(drop=True)
)

final_corrected_display = (
    final_corrected_comparison.copy()
)

metric_columns = [
    "MAE_Validation",
    "MSE_Validation",
    "R2_Validation",
    "MAE_Test",
    "MSE_Test",
    "R2_Test"
]

final_corrected_display[metric_columns] = (
    final_corrected_display[metric_columns]
    .round(4)
)

print("\n" + "=" * 100)
print("FINAL CORRECTED VALIDATION AND TEST MODEL COMPARISON")
print("=" * 100)

display(final_corrected_display)


# ------------------------------------------------------------
# 10. BEST OVERALL MODEL
# ------------------------------------------------------------

best_final_f = final_corrected_comparison.iloc[0]

print("=" * 70)
print("BEST OVERALL MODEL BASED ON TEST PERFORMANCE")
print("=" * 70)

print("Model:", best_final_f["Model"])

print(
    f"Test MAE: "
    f"{best_final_f['MAE_Test']:.4f}"
)

print(
    f"Test MSE: "
    f"{best_final_f['MSE_Test']:.4f}"
)

print(
    f"Test R²: "
    f"{best_final_f['R2_Test']:.4f}"
)


# ------------------------------------------------------------
# 11. SAVE FINAL TABLE
# ------------------------------------------------------------

final_corrected_display.to_csv(
    "../reports/final_model_comparison.csv",
    index=False
)

print(
    "\nFinal corrected comparison table saved."
)


# ------------------------------------------------------------
# 12. FINAL R2 CHART
# ------------------------------------------------------------

plt.figure(figsize=(12, 7))

plt.bar(
    final_corrected_display["Model"],
    final_corrected_display["R2_Test"]
)

plt.title(
    "Final Corrected Test R² Comparison"
)

plt.xlabel("Regression Model")
plt.ylabel("R² Score")

plt.xticks(
    rotation=35,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    "../images/final_corrected_test_r2.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 13. FINAL MAE CHART
# ------------------------------------------------------------

plt.figure(figsize=(12, 7))

plt.bar(
    final_corrected_display["Model"],
    final_corrected_display["MAE_Test"]
)

plt.title(
    "Final Corrected Test MAE Comparison"
)

plt.xlabel("Regression Model")
plt.ylabel("Mean Absolute Error (MAE)")

plt.xticks(
    rotation=35,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    "../images/final_corrected_test_mae.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nFINAL CORRECTED ANALYSIS COMPLETED.")
```

## Conclusion

This project evaluated multiple regression models to predict Airbnb listing prices using property, location, availability, booking, and review-related features. The dataset was divided into 70% training, 15% validation, and 15% test sets so that model development and final evaluation were performed on separate portions of the data.

Five required regression models were trained and evaluated: Linear Regression, Decision Tree Regressor, Random Forest Regressor, Gradient Boosting Regressor, and K-Nearest Neighbors Regressor. An average ensemble using the top three validation models and a Bayesian Ridge model were also evaluated. Model performance was compared using Mean Absolute Error (MAE), Mean Squared Error (MSE), and R².

An important issue was identified during model validation. The original results produced an unusually high R² of approximately 0.995. Further investigation showed that `service_fee` had a correlation of approximately 0.998 with `price`, indicating that it contained information closely derived from the target variable. The engineered `fee_to_price_ratio` variable also contained price information. Both variables were therefore removed from the predictor set to prevent target leakage. All models were retrained and reevaluated after this correction.

After correcting the target leakage, the **Random Forest Regressor produced the strongest overall performance**. On the validation set, Random Forest achieved an MAE of **224.8594**, MSE of **76,542.1309**, and R² of **0.3031**. On the unseen test set, it achieved an MAE of **223.6204**, MSE of **76,230.0658**, and R² of **0.3022**.

The Random Forest test R² of 0.3022 indicates that the model explained approximately **30.22% of the variation in Airbnb listing prices** in the test data. Its validation R² of 0.3031 was very close to its test R² of 0.3022. This consistency suggests that the model generalized reasonably well and did not experience a large decline when applied to unseen data.

The **Average Ensemble of the top three models** ranked second based on test R². It achieved a test MAE of **264.3793**, MSE of **94,434.8006**, and R² of **0.1356**. Although ensemble methods can sometimes improve predictions by combining several models, the ensemble did not outperform Random Forest in this project. Averaging predictions from weaker models reduced the predictive advantage of the strongest model.

The **Gradient Boosting Regressor** achieved a test R² of **0.0029**, while **Bayesian Ridge** and **Linear Regression** produced test R² values close to zero. These results suggest that simple linear relationships were not sufficient to explain Airbnb price variation using the available predictors.

K-Nearest Neighbors achieved a test R² of **-0.1873**, while the Decision Tree Regressor produced a test R² of **-0.3191**. A negative R² indicates that these models performed worse than predicting the mean target value for the test observations. The single Decision Tree also produced considerably higher MSE than Random Forest, demonstrating the advantage of combining many decision trees through the Random Forest approach.

Overall, **Random Forest Regressor was selected as the final model** because it achieved the highest test R² and lowest test MAE and MSE among the evaluated approaches. Its stronger performance suggests that nonlinear relationships and interactions between Airbnb listing characteristics are important for price prediction.

The project also demonstrates the importance of checking for target leakage rather than relying only on high model scores. The original results appeared extremely strong, but further analysis showed that price-related information was leaking into the predictors. Removing those variables resulted in lower but more realistic performance. This correction produced a more reliable evaluation of how well the models can predict Airbnb prices from information that would reasonably be available without knowing the target price.
